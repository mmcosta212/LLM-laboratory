Create a free plane on https://app.tavily.com/home. In this way, it is possibile to create a free key for your experiments:

Then create a local variable via command prompt (echo procedure) or insert it in a .env and shadows it using a .gitignore file. The free monthly plan has 1000 calls (1 call corresponds to 1 invoke, 18 September 2026)

#### Libraries

```python
import os
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_ollama import ChatOllama
from tavily import TavilyClient

load_dotenv()
tavily = TavilyClient(api_key=os.getenv("TAVILY_API_KEY")) # You have to obtain a Tavily key

RESEARCH_MODEL = "llama3.1:8b" # Ensure you have installed a model which support the tool-calling
```

#### Example

```python
# We make a searching tool using the decorator procedure from Langchain
@tool
def search(query: str) -> str:
    """
    Tool that searches in internet. 
    In input the query to search for.
    """
    print(f"Searching for {query}")
    return tavily.search(query=query)


# Modello locale tramite Ollama (you have to chose a model which support the tool-calling)
llm = ChatOllama(model=RESEARCH_MODEL, temperature=0)

# List of tools
tools = [search] 

# Agent (model and tools)
agent = create_agent(model=llm, tools=tools)

# Response
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Qual è l'ultima versione di Python?"
        }
    ]
})

print(response)
```
