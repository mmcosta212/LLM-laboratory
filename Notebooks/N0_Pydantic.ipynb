## Pydantic classes

```python
from pydantic import BaseModel, Field
from typing import Tuple, Optional
import math
```

### Example

```python
class GeometricInput(BaseModel):
    perim: float = Field(..., gt=0, description="Perimetro della figura")
    area: float = Field(..., gt=0, description="Area della figura")

def lato_resolver(geometric_input: GeometricInput) -> Optional[Tuple[float, float]]:
    """
    Calcola le lunghezze dei due lati di un rettangolo dati area e perimetro.
    Restituisce una tuple (lato_a, lato_b) o None se la figura non è realizzabile.
    """
    p = geometric_input.perim
    a = geometric_input.area
    s = p / 2  # Semiperimetro (somma dei due lati: x + y = s)
    
    # Discriminante della formula risolutiva: x^2 - s*x + a = 0
    discriminante = s**2 - 4 * a
    
    if discriminante < 0:
        # Non esiste un rettangolo reale con questa combinazione di area e perimetro
        return None
    
    lato_a = (s + math.sqrt(discriminante)) / 2
    lato_b = (s - math.sqrt(discriminante)) / 2
    
    return lato_a, lato_b
```

```python
# Rettangolo con perimetro 20 e area 24 (lati attesi: 6 e 4)
dati = GeometricInput(perim=20, area=24)
risultato = lato_resolver(dati)

if risultato:
    lato_a, lato_b = risultato
    print(f"I lati del rettangolo sono {lato_a} e {lato_b}")
else:
    print("Combinazione di area e perimetro impossibile per un rettangolo reale.")
```
